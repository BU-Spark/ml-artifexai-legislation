### 1) collect all legislations data

```python
fields_to_extract = {
    "bill_number": "bill_data.number",
    "bill_url": "bill_data.url",
    "bill_title": "bill_data.title",
    "sponsor": "sponsor",
    "sponsor_name": "sponsor.name",
    "sponsor_url": "sponsor.url",
    "summary_title": "summary.title",
    "summary_text": "summary.text",
    "source_url": "metadata.source_url",
    "view_text_url": "view_text_url",
    "download_pdf_url": "download_pdf_url",
    "is_home_rule": "is_home_rule",
    "municipality": "municipality",
    "num_cosponsors": "num_cosponsors",
    "cosponsors": "cosponsors",  
    "history": "history"
}


root_path ='../../downloaded_legislation/'
json_files = os.path.join(root_path, 'state/MA/*/*/*/*/*.json') # glob pattern for json files

df = load_bill_data(json_files, fields_to_extract, with_outlier = False)
df['sponsor_name'] = df.apply(lambda row: row['sponsor_name'] if isinstance(row['sponsor'], dict) else row['sponsor'], axis=1)
df.drop(columns = ['sponsor'], inplace=True)
df
```

Removing rows with missing sponsor data

```python
df = df[df['sponsor_name'] != "NONE"]
df = df[df['sponsor_name'] != ""]
df
```

```python
legislators = pd.read_csv('../data/raw/processed_legislators.csv')
legislators
```

#### Find new sponsors and add them to legislations

```python
def add_new_sponsors(sp, legislators, url_col, leg_name_col):
    count = 0
    for index, row in sp.iterrows():
        if row.type == 'committees':
            continue
        legislator = legislators[legislators['url'] == row[url_col]]
        if  legislator.empty:
            last_name = row[leg_name_col].split(',')[0]
            first_name = row[leg_name_col].split(',')[1].strip()
            new_data = {'url':	row[url_col],
                        'First Name':	 first_name,
                        'Last Name':	 last_name,
                        'District':      None,
                        'Party':         None, 
                        'committee_ids': None}
            legislators = pd.concat([legislators, pd.DataFrame([new_data])], ignore_index=True)
            count +=1

    print(f"Number of sponsors not found in legislators: {count}")
    return legislators

sponsors = df.groupby(['sponsor_name', 'sponsor_url']).size().sort_values(ascending=False)
sponsors =  sponsors.reset_index()
sponsors['type' ] = sponsors.sponsor_url.apply(lambda url: 'committees' if 'committees' in url.lower() else 'legislator')

legislators =  add_new_sponsors(sponsors, legislators, 'sponsor_url', 'sponsor_name')
legislators
```

#### Find new cosponsors and add them to legislations

```python
# Find all cosponsors
all_cosponsors = []

for cosponsors in df.cosponsors:
    if isinstance(cosponsors, list):
        for cosponsor in cosponsors:
            url = split_url(cosponsor['url'])[0]
            all_cosponsors.append({
                "cosponsor_name": cosponsor['name'],
                "cosponsor_url": url ,
                "cosponsor_district": cosponsor['district'],
                "type": 'committees' if 'committees' in url.lower() else 'legislator',
            })
        # all_cosponsors.extend(cosponsors)
cosponsors_df = pd.DataFrame((all_cosponsors))
cosponsors_df = cosponsors_df.drop_duplicates()

cosponsor_dist_freq = (
    cosponsors_df.groupby(['cosponsor_name', 'cosponsor_url'])['cosponsor_district']
    .agg(['nunique', lambda x: list(x.unique())])  # Count unique values and also collect them
    .sort_values(by='nunique', ascending=False)
    .rename(columns={'nunique': 'distinct_count', '<lambda_0>': 'distinct_values'})
)

cosponsor_dist_freq.distinct_values = cosponsor_dist_freq.distinct_values.apply(lambda x: [y for y in x if y != ''])
cosponsor_dist_freq['distinct_count'] = cosponsor_dist_freq.distinct_values.apply(lambda x: len(x))
# if distinct_count ==0 then distinct_values = None
cosponsor_dist_freq['distinct_values'] = cosponsor_dist_freq.distinct_values.apply(lambda x: None if len(x) == 0 else x)\


# sort by distinct_count
cosponsor_dist_freq = cosponsor_dist_freq.sort_values(by='distinct_count', ascending=False)
cosponsor_dist_freq.reset_index(inplace=True)
cosponsor_dist_freq['type'] = cosponsor_dist_freq.cosponsor_url.apply(lambda url:'committees' if 'committees' in url.lower() else 'legislator' )

cosponsor_dist_freq
    
```

```python
def add_new_cosponsors(sp, legislators, url_col, leg_name_col):
    count = 0
    for index, row in sp.iterrows():
        if row.type == 'committees':
            continue
        legislator = legislators[legislators['url'] == row[url_col]]
        if  legislator.empty:
            last_name = row[leg_name_col].split(' ')[-1]
            first_name = row[leg_name_col].split(' ')[0]
            if row['distinct_count'] == 0:
                District = None
            elif len(row['distinct_values']) == 1:
                District = row['distinct_values'][0]
            else:
                District = row['distinct_values']
            new_data = {'url':	row[url_col],
                        'First Name':	 first_name,
                        'Last Name':	 last_name,
                        'District':     District,
                        'Party':         None, 
                        'committee_ids': None}
            legislators = pd.concat([legislators, pd.DataFrame([new_data])], ignore_index=True)
            count +=1

    print(f"Number of sponsors not found in legislators: {count}")
    return legislators
legislators =  add_new_cosponsors(cosponsor_dist_freq, legislators, 'cosponsor_url', 'cosponsor_name')
legislators
```

#### Fill missing District values

```python
# fill_district
for index, row in legislators.iterrows():
    if pd.isna(row['District']):
        if row['url'] in cosponsor_dist_freq['cosponsor_url'].values:
            district = cosponsor_dist_freq[cosponsor_dist_freq['cosponsor_url'] == row['url']]['distinct_values'].values[0]
            if isinstance(district, list) and len(district) > 0:
                district = district
            else:
                district = None
            legislators.at[index, 'District'] = district
        else:
            legislators.at[index, 'District'] = None
```

### Final dataset corrections

```python
all_data = []
count = 0
for cosponsors in df.cosponsors:
    count += 1
    metadata = [] 
    if isinstance(cosponsors, list):
        if len(cosponsors) == 0:
            all_data.append(cosponsors)
            continue
        for cosponsor in cosponsors:
            if 'committees' in cosponsor['url'].lower():
                metadata.append(cosponsor)
                continue
            cos_data = cosponsor.copy()
            url = split_url(cosponsor['url'])[0]
            if cosponsor['district'] == '':
                districts = legislators[legislators['url'] == url]['District']
                if len(districts) > 0:
                    district = districts.values[0]
                else:
                    district = None
            else:

                district = cosponsor['district']
            # urls.append(str([url, district]))
            # print([url, district])
            if district is None:
                district = ''
            elif isinstance(district, list) and len(district) == 1:
                district = district[0]
            cos_data['district'] = district
            metadata.append(cos_data)

    if len(metadata) != len(cosponsors):
        print(metadata)
    
    all_data.append(metadata)

len(all_data), count
```

```python
# fill empty cosponsors's districts
df['cosponsors'] = all_data
```

```python
# correct home_rule 
df.is_home_rule = df.municipality.apply(lambda x: pd.notna(x))
```
