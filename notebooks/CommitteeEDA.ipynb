Look at list of committee-legislator pairs


```python
import pandas as pd
import os

file_path = os.path.join('..', 'data', 'raw', 'committees_members.csv')

df = pd.read_csv(file_path)


df
```

Retrieve list of unique committee names

```python
committee_df = pd.DataFrame()
committee_df = df['committee_name'].dropna().unique()
committee_df = pd.DataFrame(committee_df, columns=['Committee'])
committee_df
```

```python
committee_df.to_csv('committees.csv', index=True)
```

```python
legislator_file_path = os.path.join('..', 'data', 'raw', 'legislators.csv')

# Read the CSV file into a DataFrame
legislator_df = pd.read_csv(legislator_file_path)

legislator_df
```

Add committee ids to legislators where the legislator belongs to each committee added

```python
# legislator_df['committee_ids'] = []
full_list = []
for index, row in legislator_df.iterrows():
    #print(row['url'])
    url = row['url']
    result = df[df['profile_url']==url] 
    #print(result)
    names = result['committee_name']

    committee_id_list = []
    for name in names:
        # print(name)
        committee_result = committee_df[committee_df['Committee']==name]
        # print(committee_result.index[0])
        committee_index = committee_result.index[0]
        committee_id_list.append(committee_index)
    
    full_list += [committee_id_list]
    # print(index, committee_id_list)
    # legislator_df.loc[index, 'committee_ids'] = committee_id_list
print(full_list)

legislator_df['committee_ids'] = full_list

legislator_df
```

```python
legislator_df.to_csv('processed_legislators.csv', index=False)
```
