Now we want to standardize and classify sponsors as either a person or committee/commission. In order to do so, we will be using spaCy to first classify which sponsor names are a real person.

We must do pip3 install spacy and python3 -m spacy download en_core_web_sm for everything to work

```python
import spacy

nlp = spacy.load("en_core_web_sm")


def preprocess_name(name):
    if "," in name:
        parts = name.split(", ")
        
        name = f"{parts[-1]} {parts[0]}"
    return name

def classify_sponsor(sponsor):

    processed = preprocess_name(sponsor)
    doc = nlp(processed)

    for ent in doc.ents:
        if ent.label_ == "PERSON":
            return "Person"
        else:
            return "Group"
    return "Other"


sponsors = pd.DataFrame()

sponsors['sponsor'] = df['sponsor']
sponsors['person_label'] = df['sponsor'].apply(classify_sponsor)

sponsors[sponsors['person_label'] == 'Group']
```
