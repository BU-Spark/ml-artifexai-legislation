Import Bills data

```python
import os
import pandas as pd


billsPath =  os.path.join('..', 'data', 'raw', 'bills.csv')
billsDF = pd.read_csv(billsPath, engine='python')
billsDF = billsDF[billsDF['bill_url'].notnull()]
billsDF
```

Checking to see if bills.csv is up to date with corrected_data_sample.csv

```python
billsDF["cosponsors"]

correctedPath =  os.path.join('..', 'data', 'raw', 'corrected_data.csv')
correctedDF = pd.read_csv(correctedPath, engine='python')
correctedDF = correctedDF[correctedDF['bill_url'].notnull()]
correctedDF

#(billsDF["cosponsors"] == correctedDF["cosponsors"]).all
# correctedDF[correctedDF["sponsor_name"] == "NONE"]
```

Merging corrected data with bill data 

```python
merged_df = pd.merge(billsDF, correctedDF, on='bill_url', suffixes=('', '_merged_df'), how='left')
merged_df["cosponsors"] = merged_df["cosponsors_merged_df"]
merged_df.drop(columns = ["cosponsors_merged_df","bill_number_merged_df","sponsor_name","sponsor_url_merged_df","bill_title","summary_title","summary_text","source_url","view_text_url_merged_df","download_pdf_url_merged_df","is_home_rule_merged_df","municipality_merged_df","num_cosponsors_merged_df","history_merged_df"], inplace=True)

pd.set_option('display.max_columns', None)
merged_df = merged_df[merged_df['bill_url'].notnull()]
merged_df[merged_df['bill_url'].isna()]
merged_df = merged_df.reset_index(drop=True)
merged_df
```

Ensuring there are no duplicates

```python


merged_df = merged_df.drop_duplicates(subset='bill_url')
duplicates = merged_df[merged_df['bill_url'].duplicated(keep=False)]

# Display the rows with duplicate values in the 'cosponsors' column
print(duplicates)
```

```python
merged_df.to_csv('../data/raw/final_bill_data.csv', index=False, quotechar='"', quoting=2, encoding='utf-8')
```
